```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

#load data
df = pd.read_csv("calon_tni.csv")

# cetak informasi dari data frame
print("========================== Data Calon TNI ==========================")
print("=" * 68)
print(df)
print("=" * 68)

print()
print()

# cetak statistik deskriptif
print("=" * 68)
print("======================= Statistik Deskriptif =======================")
print(df.describe())

print()
print()

# deteksi masalah dalam data / missing value
print("=" * 42)
print("====== Deteksi Masalah Missing Value =====")
print(df.isnull().sum())

print()
print()

# data anomali atau data yang tidak wajar
print("=" * 47)
print("====== Deteksi Data Tidak Wajar / Anomali =====")
gaji_maks = df["gaji"].max()
gaji_mini = df["gaji"].min()
rata_rata = df["gaji"].mean()
nama_kosong = df["nama"].isnull().sum()
alamat_kosong = df["alamat"].isnull().sum()
tinggi_kosong = df["tinggi"].isnull().sum()
jenis_kelamin_kosong = df["jenis_kelamin"].isnull().sum()
gaji_kosong = df["gaji"].isnull().sum()
print(f"Gaji Paling Tinggi: {gaji_maks:,}".replace(",", "."))
print(f"Gaji Paling Rendah: {gaji_mini:,}".replace(",", "."))
print(f"Rata-Rata Gaji: {rata_rata:,}".replace(",", "."))
print(f"Nama kosong: {nama_kosong} data")
print(f"Alamat kosong: {alamat_kosong} data")
print(f"Tinggi badan kosong: {tinggi_kosong} data")
print(f"Jenis kelamin kosong: {jenis_kelamin_kosong} data")
print(f"Gaji kosong: {gaji_kosong} data")
```
