```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

#load data
df = pd.read_csv("calon_tni.csv")

# cetak informasi dari data frame
print("========================== Data Calon TNI ==========================")
print("=" * 68)
print(df)
print("=" * 68)

print()
print()

# cetak statistik deskriptif
print("=" * 68)
print("======================= Statistik Deskriptif =======================")
print(df.describe())

print()
print()

# deteksi masalah dalam data / missing value
print("=" * 42)
print("====== Deteksi Masalah Missing Value =====")
print(df.isnull().sum())

print()
print()

# data anomali atau data yang tidak wajar
print("=" * 47)
print("====== Deteksi Data Tidak Wajar / Anomali =====")
gaji_maks = df["gaji"].max()
gaji_mini = df["gaji"].min()
rata_rata = df["gaji"].mean()
nama_kosong = df["nama"].isnull().sum()
alamat_kosong = df["alamat"].isnull().sum()
tinggi_kosong = df["tinggi"].isnull().sum()
jenis_kelamin_kosong = df["jenis_kelamin"].isnull().sum()
gaji_kosong = df["gaji"].isnull().sum()
print(f"Gaji Paling Tinggi: {gaji_maks:,}".replace(",", "."))
print(f"Gaji Paling Rendah: {gaji_mini:,}".replace(",", "."))
print(f"Rata-Rata Gaji: {rata_rata:,}".replace(",", "."))
print(f"Nama kosong: {nama_kosong} data")
print(f"Alamat kosong: {alamat_kosong} data")
print(f"Tinggi badan kosong: {tinggi_kosong} data")
print(f"Jenis kelamin kosong: {jenis_kelamin_kosong} data")
print(f"Gaji kosong: {gaji_kosong} data")

print()
print()

# DETEKSI OUTLIER UNTUK GAJI MENGGUNAKAN IQR METHOD
print("=" * 47)
print("====== Deteksi Outlier Gaji (IQR Method) =====")

def detect_outliers_iqr(data, column_name):
    Q1 = data[column_name].quantile(0.25)
    Q3 = data[column_name].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = data[(data[column_name] < lower_bound) | (data[column_name] > upper_bound)]
    
    print(f"Statistik untuk {column_name}:")
    print(f"Q1 (25%): {Q1:,.0f}".replace(",", "."))
    print(f"Q3 (75%): {Q3:,.0f}".replace(",", "."))
    print(f"IQR: {IQR:,.0f}".replace(",", "."))
    print(f"Batas bawah: {lower_bound:,.0f}".replace(",", "."))
    print(f"Batas atas: {upper_bound:,.0f}".replace(",", "."))
    print(f"Jumlah outlier: {len(outliers)}")
    
    if len(outliers) > 0:
        print("Data outlier:")
        for idx, row in outliers.iterrows():
            nama = row['nama'] if pd.notna(row['nama']) else 'Nama Kosong'
            print(f"  - {nama}: {row[column_name]:,}".replace(",", "."))
    else:
        print("  Tidak ada outlier")
    print()

# Deteksi outlier untuk gaji
detect_outliers_iqr(df, 'gaji')

print()
print()

# VISUALISASI BOX PLOT UNTUK GAJI
print("=" * 47)
print("====== Visualisasi Box Plot Gaji =====")

plt.figure(figsize=(8, 6))

# PERBAIKAN: Menggunakan tick_labels sebagai ganti labels
boxplot = plt.boxplot(df['gaji'], patch_artist=True, tick_labels=['Gaji'])

# Warna untuk box plot
colors = ['lightblue']
for patch, color in zip(boxplot['boxes'], colors):
    patch.set_facecolor(color)

plt.title('Box Plot Distribusi Gaji Calon TNI', fontsize=14, fontweight='bold')
plt.ylabel('Gaji (Rupiah)', fontsize=12)

# Format y-axis untuk gaji
plt.gca().yaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: f'{x:,.0f}'.replace(",", ".")))

# Tambahkan grid
plt.grid(True, alpha=0.3)

# Tambahkan annotasi untuk statistik penting
Q1 = df['gaji'].quantile(0.25)
Q3 = df['gaji'].quantile(0.75)
median = df['gaji'].median()
mean = df['gaji'].mean()

plt.axhline(y=median, color='red', linestyle='--', alpha=0.7, label=f'Median: {median:,.0f}'.replace(",", "."))
plt.axhline(y=Q1, color='orange', linestyle='--', alpha=0.7, label=f'Q1: {Q1:,.0f}'.replace(",", "."))
plt.axhline(y=Q3, color='green', linestyle='--', alpha=0.7, label=f'Q3: {Q3:,.0f}'.replace(",", "."))
plt.axhline(y=mean, color='blue', linestyle='--', alpha=0.7, label=f'Rata-rata: {mean:,.0f}'.replace(",", "."))

plt.legend()

plt.tight_layout()
plt.show()

print("Analisis selesai!")
```
