```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

# Load data
df = pd.read_csv('titanic.csv')

# Data preprocessing
print("Informasi dataset:")
print(df.info())
print("\n5 data pertama:")
print(df.head())

# Cek missing values
print("\nMissing values:")
print(df.isnull().sum())

# Handle missing values - isi age dengan median
df['age'].fillna(df['age'].median(), inplace=True)

# Encode categorical variables
le_sex = LabelEncoder()
df['sex_encoded'] = le_sex.fit_transform(df['sex'])

print("\nDistribusi target (Survived):")
print(df['survived'].value_counts())

# Fungsi untuk evaluasi model
def evaluate_model(y_true, y_pred, model_name):
    accuracy = accuracy_score(y_true, y_pred)
    print(f"\n{model_name}")
    print(f"Akurasi: {accuracy:.4f}")
    print("\nClassification Report:")
    print(classification_report(y_true, y_pred))
    
    # Confusion Matrix
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(6, 4))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
    plt.title(f'Confusion Matrix - {model_name}')
    plt.ylabel('Actual')
    plt.xlabel('Predicted')
    plt.show()
    
    return accuracy

# Fungsi untuk train dan test model KNN
def knn_classification(features, target, test_size=0.2, random_state=42, k=5):
    # Split data
    X_train, X_test, y_train, y_test = train_test_split(
        features, target, test_size=test_size, random_state=random_state, stratify=target
    )
    
    # Standardisasi fitur
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)
    
    # Train KNN
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train_scaled, y_train)
    
    # Predict
    y_pred = knn.predict(X_test_scaled)
    
    return y_test, y_pred, knn

# MODEL 1: Fitur (Passenger Class, Sex, Age)
print("="*60)
print("MODEL 1: Fitur (Passenger Class, Sex, Age)")
print("="*60)

features1 = df[['pclass', 'sex_encoded', 'age']]
target = df['survived']

y_test1, y_pred1, knn1 = knn_classification(features1, target)
accuracy1 = evaluate_model(y_test1, y_pred1, "Model 1: Pclass, Sex, Age")

# MODEL 2: Fitur (Sex, Age, Fare)
print("="*60)
print("MODEL 2: Fitur (Sex, Age, Fare)")
print("="*60)

features2 = df[['sex_encoded', 'age', 'fare']]
y_test2, y_pred2, knn2 = knn_classification(features2, target)
accuracy2 = evaluate_model(y_test2, y_pred2, "Model 2: Sex, Age, Fare")

# MODEL 3: Fitur (Passenger Class, Fare)
print("="*60)
print("MODEL 3: Fitur (Passenger Class, Fare)")
print("="*60)

features3 = df[['pclass', 'fare']]
y_test3, y_pred3, knn3 = knn_classification(features3, target)
accuracy3 = evaluate_model(y_test3, y_pred3, "Model 3: Pclass, Fare")

# Perbandingan akurasi semua model
print("="*60)
print("PERBANDINGAN HASIL SEMUA MODEL")
print("="*60)

models = ['Pclass, Sex, Age', 'Sex, Age, Fare', 'Pclass, Fare']
accuracies = [accuracy1, accuracy2, accuracy3]

comparison_df = pd.DataFrame({
    'Model': models,
    'Accuracy': accuracies
})

print(comparison_df)

# Visualisasi perbandingan akurasi
plt.figure(figsize=(10, 6))
bars = plt.bar(models, accuracies, color=['skyblue', 'lightcoral', 'lightgreen'])
plt.title('Perbandingan Akurasi Model KNN dengan Fitur Berbeda')
plt.ylabel('Akurasi')
plt.ylim(0, 1)

# Tambah nilai di atas bar
for bar, accuracy in zip(bars, accuracies):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.01, 
             f'{accuracy:.4f}', ha='center', va='bottom')

plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

# Analisis nilai K optimal untuk model terbaik
print("="*60)
print("ANALISIS NILAI K OPTIMAL UNTUK MODEL TERBAIK")
print("="*60)

best_model_idx = np.argmax(accuracies)
best_features = [features1, features2, features3][best_model_idx]
best_model_name = models[best_model_idx]

print(f"Mencari nilai K optimal untuk model: {best_model_name}")

# Cari K optimal
k_values = range(1, 21)
accuracies_k = []

X_train, X_test, y_train, y_test = train_test_split(
    best_features, target, test_size=0.2, random_state=42, stratify=target
)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

for k in k_values:
    knn_temp = KNeighborsClassifier(n_neighbors=k)
    knn_temp.fit(X_train_scaled, y_train)
    y_pred_temp = knn_temp.predict(X_test_scaled)
    accuracy_temp = accuracy_score(y_test, y_pred_temp)
    accuracies_k.append(accuracy_temp)

# Plot akurasi vs K
plt.figure(figsize=(12, 6))
plt.plot(k_values, accuracies_k, marker='o', linestyle='-', color='blue')
plt.title(f'Akurasi vs Nilai K untuk Model: {best_model_name}')
plt.xlabel('Nilai K')
plt.ylabel('Akurasi')
plt.grid(True, alpha=0.3)
plt.xticks(k_values)
plt.tight_layout()
plt.show()

# Nilai K optimal
optimal_k = k_values[np.argmax(accuracies_k)]
print(f"\nNilai K optimal: {optimal_k} dengan akurasi: {max(accuracies_k):.4f}")

# Train model final dengan K optimal
print(f"\nModel Final dengan K={optimal_k} untuk {best_model_name}:")
knn_final = KNeighborsClassifier(n_neighbors=optimal_k)
knn_final.fit(X_train_scaled, y_train)
y_pred_final = knn_final.predict(X_test_scaled)

final_accuracy = evaluate_model(y_test, y_pred_final, f"Model Final (K={optimal_k})")

# Contoh prediksi untuk data baru
print("="*60)
print("CONTOH PREDIKSI UNTUK DATA BARU")
print("="*60)

# Data baru untuk prediksi
new_data = pd.DataFrame({
    'pclass': [1, 3, 2],
    'sex_encoded': [0, 1, 0],  # 0: female, 1: male
    'age': [25, 30, 8],
    'fare': [100, 10, 25]
})

print("Data baru untuk prediksi:")
print(new_data)

# Prediksi dengan model terbaik
if best_model_name == 'Pclass, Sex, Age':
    new_features = new_data[['pclass', 'sex_encoded', 'age']]
elif best_model_name == 'Sex, Age, Fare':
    new_features = new_data[['sex_encoded', 'age', 'fare']]
else:  # 'Pclass, Fare'
    new_features = new_data[['pclass', 'fare']]

# Standardisasi dan prediksi
new_features_scaled = scaler.transform(new_features)
predictions = knn_final.predict(new_features_scaled)
probabilities = knn_final.predict_proba(new_features_scaled)

print("\nHasil Prediksi:")
for i, (pred, prob) in enumerate(zip(predictions, probabilities)):
    status = "Selamat" if pred == 1 else "Tidak Selamat"
    prob_survived = prob[1] * 100
    print(f"Data {i+1}: {status} (Probabilitas selamat: {prob_survived:.2f}%)")

# Analisis feature importance (untuk insight)
print("\n" + "="*60)
print("ANALISIS PENGARUH FITUR")
print("="*60)

# Korelasi dengan target
correlation_with_target = df[['pclass', 'age', 'fare', 'sex_encoded', 'survived']].corr()['survived'].drop('survived')
print("Korelasi fitur dengan target (Survived):")
print(correlation_with_target.sort_values(ascending=False))

# Visualisasi korelasi
plt.figure(figsize=(8, 6))
sns.heatmap(df[['pclass', 'age', 'fare', 'sex_encoded', 'survived']].corr(), 
            annot=True, cmap='coolwarm', center=0)
plt.title('Korelasi antara Fitur dan Target')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report

# Load data
df = pd.read_csv('titanic.csv')

# Data preprocessing
print("Informasi dataset:")
print(f"Total data: {len(df)}")
print(f"Yang selamat: {df['survived'].sum()} ({df['survived'].mean()*100:.1f}%)")
print(f"Yang tidak selamat: {len(df) - df['survived'].sum()} ({100 - df['survived'].mean()*100:.1f}%)")

# PERBAIKAN: Handle missing values - isi age dengan median (tanpa inplace)
df['age'] = df['age'].fillna(df['age'].median())

# Encode categorical variables
le_sex = LabelEncoder()
df['sex_encoded'] = le_sex.fit_transform(df['sex'])

# Fungsi untuk train dan test model KNN
def knn_classification(features, target, test_size=0.2, random_state=42, k=5):
    # Split data
    X_train, X_test, y_train, y_test = train_test_split(
        features, target, test_size=test_size, random_state=random_state, stratify=target
    )
    
    # Standardisasi fitur
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)
    
    # Train KNN
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train_scaled, y_train)
    
    # Predict
    y_pred = knn.predict(X_test_scaled)
    
    return y_test, y_pred, knn, scaler

# MODEL 1: Fitur (Passenger Class, Sex, Age)
print("\n" + "="*60)
print("MODEL 1: Fitur (Passenger Class, Sex, Age)")
print("="*60)

features1 = df[['pclass', 'sex_encoded', 'age']]
target = df['survived']

y_test1, y_pred1, knn1, scaler1 = knn_classification(features1, target)
accuracy1 = accuracy_score(y_test1, y_pred1)
print(f"Akurasi: {accuracy1:.4f}")

# MODEL 2: Fitur (Sex, Age, Fare)
print("\n" + "="*60)
print("MODEL 2: Fitur (Sex, Age, Fare)")
print("="*60)

features2 = df[['sex_encoded', 'age', 'fare']]
y_test2, y_pred2, knn2, scaler2 = knn_classification(features2, target)
accuracy2 = accuracy_score(y_test2, y_pred2)
print(f"Akurasi: {accuracy2:.4f}")

# MODEL 3: Fitur (Passenger Class, Fare)
print("\n" + "="*60)
print("MODEL 3: Fitur (Passenger Class, Fare)")
print("="*60)

features3 = df[['pclass', 'fare']]
y_test3, y_pred3, knn3, scaler3 = knn_classification(features3, target)
accuracy3 = accuracy_score(y_test3, y_pred3)
print(f"Akurasi: {accuracy3:.4f}")

# Perbandingan akurasi semua model
print("\n" + "="*60)
print("PERBANDINGAN HASIL SEMUA MODEL")
print("="*60)

models = ['Pclass, Sex, Age', 'Sex, Age, Fare', 'Pclass, Fare']
accuracies = [accuracy1, accuracy2, accuracy3]

comparison_df = pd.DataFrame({
    'Model': models,
    'Accuracy': accuracies
})

print(comparison_df)

# Tentukan model terbaik
best_model_idx = np.argmax(accuracies)
best_model_name = models[best_model_idx]
best_accuracy = accuracies[best_model_idx]

print(f"\nModel terbaik: {best_model_name}")
print(f"Akurasi terbaik: {best_accuracy:.4f}")

# Analisis nilai K optimal untuk model terbaik
print("\n" + "="*60)
print("ANALISIS NILAI K OPTIMAL UNTUK MODEL TERBAIK")
print("="*60)

best_features = [features1, features2, features3][best_model_idx]

# Cari K optimal
k_values = range(1, 21)
accuracies_k = []

X_train, X_test, y_train, y_test = train_test_split(
    best_features, target, test_size=0.2, random_state=42, stratify=target
)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

for k in k_values:
    knn_temp = KNeighborsClassifier(n_neighbors=k)
    knn_temp.fit(X_train_scaled, y_train)
    y_pred_temp = knn_temp.predict(X_test_scaled)
    accuracy_temp = accuracy_score(y_test, y_pred_temp)
    accuracies_k.append(accuracy_temp)

# Nilai K optimal
optimal_k = k_values[np.argmax(accuracies_k)]
optimal_accuracy = max(accuracies_k)

print(f"Nilai K optimal: {optimal_k}")
print(f"Akurasi dengan K={optimal_k}: {optimal_accuracy:.4f}")

# Train model final dengan K optimal
print(f"\nModel Final dengan K={optimal_k} untuk {best_model_name}:")
knn_final = KNeighborsClassifier(n_neighbors=optimal_k)
knn_final.fit(X_train_scaled, y_train)
y_pred_final = knn_final.predict(X_test_scaled)

final_accuracy = accuracy_score(y_test, y_pred_final)
print(f"Akurasi model final: {final_accuracy:.4f}")

# Contoh prediksi untuk data baru
print("\n" + "="*60)
print("CONTOH PREDIKSI UNTUK DATA BARU")
print("="*60)

# Data baru untuk prediksi
new_data = pd.DataFrame({
    'pclass': [1, 3, 2, 1, 3],
    'sex_encoded': [0, 1, 0, 1, 0],  # 0: female, 1: male
    'age': [25, 30, 8, 45, 5],
    'fare': [100, 10, 25, 200, 15]
})

print("Data baru untuk prediksi:")
print(new_data)

# Prediksi dengan model terbaik
if best_model_name == 'Pclass, Sex, Age':
    new_features = new_data[['pclass', 'sex_encoded', 'age']]
    print("Fitur yang digunakan: Passenger Class, Sex, Age")
elif best_model_name == 'Sex, Age, Fare':
    new_features = new_data[['sex_encoded', 'age', 'fare']]
    print("Fitur yang digunakan: Sex, Age, Fare")
else:  # 'Pclass, Fare'
    new_features = new_data[['pclass', 'fare']]
    print("Fitur yang digunakan: Passenger Class, Fare")

# Standardisasi dan prediksi
new_features_scaled = scaler.transform(new_features)
predictions = knn_final.predict(new_features_scaled)
probabilities = knn_final.predict_proba(new_features_scaled)

print("\nHasil Prediksi Survival:")
print("-" * 50)
for i, (pred, prob) in enumerate(zip(predictions, probabilities)):
    status = "SELAMAT" if pred == 1 else "TIDAK SELAMAT"
    prob_survived = prob[1] * 100
    gender = "Perempuan" if new_data.iloc[i]['sex_encoded'] == 0 else "Laki-laki"
    pclass = new_data.iloc[i]['pclass']
    age = new_data.iloc[i]['age']
    
    print(f"Penumpang {i+1}:")
    print(f"  - Kelas: {pclass}")
    print(f"  - Gender: {gender}")
    print(f"  - Usia: {age} tahun")
    print(f"  - Status: {status}")
    print(f"  - Probabilitas selamat: {prob_survived:.2f}%")
    print("-" * 30)

# Analisis korelasi fitur
print("\n" + "="*60)
print("ANALISIS KORELASI FITUR DENGAN SURVIVAL")
print("="*60)

correlation_with_target = df[['pclass', 'age', 'fare', 'sex_encoded', 'survived']].corr()['survived'].drop('survived')
print("Korelasi fitur dengan survival:")
for feature, corr in correlation_with_target.items():
    if feature == 'sex_encoded':
        feature_name = 'sex'
    else:
        feature_name = feature
    print(f"  {feature_name}: {corr:.4f}")

# Statistik survival
print("\n" + "="*60)
print("STATISTIK SURVIVAL BERDASARKAN GENDER DAN KELAS")
print("="*60)

print("\nSurvival rate berdasarkan Gender:")
gender_survival = df.groupby('sex')['survived'].mean()
for gender, rate in gender_survival.items():
    print(f"  {gender}: {rate:.2%}")

print("\nSurvival rate berdasarkan Kelas:")
class_survival = df.groupby('pclass')['survived'].mean()
for pclass, rate in class_survival.items():
    print(f"  Kelas {pclass}: {rate:.2%}")

print("\nRata-rata usia:")
print(f"  Selamat: {df[df['survived'] == 1]['age'].mean():.1f} tahun")
print(f"  Tidak selamat: {df[df['survived'] == 0]['age'].mean():.1f} tahun")

print(f"\nRata-rata tarif:")
print(f"  Selamat: ${df[df['survived'] == 1]['fare'].mean():.2f}")
print(f"  Tidak selamat: ${df[df['survived'] == 0]['fare'].mean():.2f}")

# Ringkasan akhir
print("\n" + "="*60)
print("RINGKASAN HASIL PREDIKSI KNN")
print("="*60)
print(f"Model terbaik: {best_model_name}")
print(f"Akurasi terbaik: {best_accuracy:.4f}")
print(f"Nilai K optimal: {optimal_k}")
print(f"Total data training: {len(X_train)}")
print(f"Total data testing: {len(X_test)}")
```
